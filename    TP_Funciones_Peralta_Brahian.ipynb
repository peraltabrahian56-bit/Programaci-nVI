```python
# Trabajo Práctico: Funciones en Python
# Docente: Ing. Ricardo Maidana
# Fecha de asignación: 2 de septiembre de 2026
# Fecha límite de presentación: 10 de septiembre de 2026
# Puntaje total: 50 puntos

# Datos del estudiante
# Nombre y apellido: Brahian Osvaldo Peralta Correa.
# Curso/Sección: Cuarto Curso Seccion B
# Fecha de entrega: 10 de septiembre de 2026

# Propósito del trabajo
# Aplicar los conceptos estudiados sobre definición de funciones, retorno y envío de valores, argumentos y parámetros, paso de objetos, argumentos indeterminados, recursividad y funciones integradas de Python.

# Ejercicio 1 — Ficha de bienvenida
def mostrar_bienvenida():
    print("Curso: Ingeniería Informática - Programación con Python")
    print("¡Bienvenido/a al Trabajo Práctico de Funciones!")
    print("Aprender funciones facilita la programación")

mostrar_bienvenida()
mostrar_bienvenida()

print("\n" + "="*50 + "\n")

# Ejercicio 2 — Calculadora de compra
def calcular_total(precio, cantidad):
    return precio * cantidad

total_cuadernos = calcular_total(18000, 3)
total_boligrafos = calcular_total(3500, 5)
total_general = total_cuadernos + total_boligrafos

print(f"Total cuadernos: Gs. {total_cuadernos}")
print(f"Total bolígrafos: Gs. {total_boligrafos}")
print(f"Total general: Gs. {total_general}")

print("\n" + "="*50 + "\n")

# Ejercicio 3 — Resumen de calificaciones
def resumen_notas(nota1, nota2, nota3):
    mayor = max(nota1, nota2, nota3)
    menor = min(nota1, nota2, nota3)
    promedio = round((nota1 + nota2 + nota3) / 3, 2)
    return mayor, menor, promedio

nota_mayor, nota_menor, promedio = resumen_notas(72, 88, 95)
print(f"Nota mayor: {nota_mayor}")
print(f"Nota menor: {nota_menor}")
print(f"Promedio: {promedio}")

print("\n" + "="*50 + "\n")

# Ejercicio 4 — Registro de estudiante
def presentar_estudiante(nombre, curso, ciudad="Caacupé"):
    print(" Ficha del estudiante")
    print(f"Nombre: {nombre}")
    print(f"Curso: {curso}")
    print(f"Ciudad: {ciudad}")
    print()

presentar_estudiante("Ana", "Tercer año")
presentar_estudiante(curso="Segundo año", nombre="Luis")
presentar_estudiante("Marcos", "Primer año", "Asunción")

print("="*50 + "\n")

# Ejercicio 5 — Alcance de variables
institucion = "Centro Educativo Juan Pablo II"

def mostrar_institucion():
    print(f"Institución: {institucion}")
    mensaje = "Laboratorio de Informática"
    print(f"Mensaje local: {mensaje}")

mostrar_institucion()
# print(mensaje) # Esto daría error porque 'mensaje' es local

print("\nExplicación: La variable `mensaje` es local porque fue creada dentro de la función. Solo existe ahí adentro. Por eso fuera da error. `institucion` es global y se puede usar en cualquier parte.\n")

print("="*50 + "\n")

# Ejercicio 6 — Lista original y lista modificada
precios = [12000, 25000, 18000, 30000]

def aplicar_descuento(lista, porcentaje):
    for i in range(len(lista)):
        lista[i] = lista[i] - (lista[i] * porcentaje / 100)
    return lista

precios_con_descuento = aplicar_descuento(precios.copy(), 10)

print(f"Lista original: {precios}")
print(f"Lista modificada: {precios_con_descuento}")

print("\nExplicación: Si enviaba `precios` sin `.copy()` se modificaba la lista original porque las listas se pasan por referencia. Con `.copy()` trabajo sobre una copia y la original queda intacta.\n")

print("="*50 + "\n")

# Ejercicio 7 — Promedio de cantidad variable de notas
def calcular_promedio(*notas):
    if len(notas) == 0:
        return "Advertencia: no se recibió ninguna nota para calcular el promedio."
    return round(sum(notas) / len(notas), 2)

print(calcular_promedio(70, 85, 90))
print(calcular_promedio(60, 75, 80, 95, 100))
print(calcular_promedio())
print(calcular_promedio(88, 92, 79, 85, 90, 95))

print("\n" + "="*50 + "\n")

# Ejercicio 8 — Comprobante dinámico
def generar_comprobante(**datos):
    print(" Comprobante")
    for clave, valor in datos.items():
        print(f"{clave}: {valor}")
    print()

generar_comprobante(numero=105, cliente="María López", concepto="Cuota mensual", monto=150000)
generar_comprobante(numero=106, cliente="Javier Invernizzi", concepto="Matrícula", monto=250000, fecha="09/09/2026", metodo_pago="Transferencia")

print("="*50 + "\n")

# Ejercicio 9 — Cuenta regresiva recursiva
def cuenta_regresiva(numero):
    if numero < 1:
        print("El número debe ser mayor o igual a 1.")
        return
    if numero == 1: # Caso base: detiene la recursividad
        print(1)
        print("¡Inicio!")
        return
    print(numero)
    cuenta_regresiva(numero - 1)

cuenta_regresiva(5)
print("---")
cuenta_regresiva(0)

print("\n" + "="*50 + "\n")

# Ejercicio 10 — Sistema de ventas escolar
productos = [
    {"nombre": "Cuaderno", "precio": 18000, "cantidad": 2},
    {"nombre": "Regla", "precio": 5000, "cantidad": 3},
    {"nombre": "Mochila", "precio": 120000, "cantidad": 1}
]

def calcular_subtotal(precio, cantidad):
    return precio * cantidad

def calcular_descuento(subtotal, porcentaje=0):
    return subtotal * porcentaje / 100

def calcular_total(*subtotales):
    return sum(subtotales)

def mostrar_factura(cliente, **detalles):
    print("========== FACTURA ==========")
    print(f"Cliente: {cliente}")
    for clave, valor in detalles.items():
        print(f"{clave}: {valor}")
    print("==============================")

def formatear_gs(monto):
    return f"Gs. {monto:,.0f}".replace(",", ".")

subtotales = [calcular_subtotal(p["precio"], p["cantidad"]) for p in productos]
subtotal_general = calcular_total(*subtotales)
descuento = calcular_descuento(subtotal_general, 10)
total_final = subtotal_general - descuento

mostrar_factura(
    "Brahian",
    Subtotal_general=formatear_gs(subtotal_general),
    Descuento_10_por_ciento=formatear_gs(descuento),
    Total_final=formatear_gs(total_final)
)

print("\n" + "="*50 + "\n")

# Reflexión final
print("1. ¿Qué ventaja ofrece dividir un programa en funciones?")
print("Dividir un programa en funciones nos permite reducir el codigo en bloques pequeños y reutilizables, facilitando asi la deteccion y correccion de errores, la lectura del codigo, evitar repetir codigo y facilitando el mantenimiento del programa.\n")

print("2. ¿Cuál fue el ejercicio más desafiante y por qué?")
print("El ejercicio más desafiante fue el del sistema de ventas escolar, ya que requirió combinar varias funciones distintas con parámetros por defecto, *args y **kwargs y coordinarlas correctamente para producir un resultado final coherente, la factura.\n")

print("3. ¿En qué situación real utilizaría una función con *args o **kwargs?")
print("Una función con *args sería útil para sumar los montos de una cantidad variable de facturas o compras, y una función con **kwargs sería útil para generar un comprobante o formulario donde los campos pueden variar según el cliente por ejemplo, algunos con descuento y otros sin descuento.")
```
