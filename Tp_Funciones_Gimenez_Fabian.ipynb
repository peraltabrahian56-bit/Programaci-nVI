### Ejercicio 3 — Resumen de calificaciones (4 puntos)

Cree `resumen_notas(nota1, nota2, nota3)` que retorne simultáneamente la nota mayor, la nota menor y el promedio redondeado a dos decimales. Desempaquete los tres valores en variables y muéstrelos.

Pruebe con las notas `72`, `88` y `95`.

**Contenido aplicado:** retorno múltiple y funciones integradas `max()`, `min()` y `round()`.

```python
# Escriba su solución aquí
def resumen_notas(nota1, nota2, nota3):
    mayor = max(nota1, nota2, nota3)
    menor = min(nota1, nota2, nota3)
    promedio = round((nota1 + nota2 + nota3) / 3, 2)
    return mayor, menor, promedio
#notas de prueba
nota_mayor, nota_menor, promedio = resumen_notas(72, 88, 95)
print(f"Nota mayor: {nota_mayor}")
print(f"Nota menor: {nota_menor}")
print(f"Promedio: {promedio}")

```

### Ejercicio 6 — Lista original y lista modificada (5 puntos)

Parta de `precios = [12000, 25000, 18000, 30000]`. Cree `aplicar_descuento(lista, porcentaje)` para modificar cada elemento de la lista descontando el porcentaje recibido.

1. Envíe una **copia** de `precios` a la función para aplicar un descuento del 10 %.
2. Muestre la lista modificada y compruebe que la lista original no cambió.
3. Explique brevemente qué habría ocurrido al enviar directamente la lista original.

**Contenido aplicado:** objetos mutables, paso de argumentos y copia de listas.

```python
precios = [12000, 25000, 18000, 30000]

# Escriba su solución aquí
def aplicar_descuento(lista, porcentaje):
    for i in range(len(lista)):
        lista[i] = lista[i] - (lista[i] * porcentaje / 100)
    return lista

precios_con_descuento = aplicar_descuento(precios.copy(), 10)

print(f"Lista original:  {precios}")
print(f"Lista modificada: {precios_con_descuento}")
```

**Explicación del estudiante:**  
Las listas son objetos mutables, y en Python los objetos mutables se pasan a las funciones por referencia. Si se hubiera enviado directamente `precios` (sin usar `.copy()`), la función habría modificado la lista original dentro del mismo espacio de memoria, ya que `lista` dentro de la función apuntaría al mismo objeto que `precios`. Al enviar una copia con `precios.copy()`, la función trabaja sobre un objeto independiente y la lista original queda intacta.

### Ejercicio 9 — Cuenta regresiva recursiva (6 puntos)

Cree una función recursiva `cuenta_regresiva(numero)` que muestre desde el número recibido hasta `1` y finalice mostrando `¡Inicio!`.

La función debe incluir:

- Un **caso base** que detenga las llamadas.
- Una **llamada recursiva** que reduzca el número.
- Una validación: si el valor inicial es menor que 1, debe mostrar un mensaje adecuado.

Pruebe con `5` y con `0`. Después identifique el caso base en un comentario.

**Contenido aplicado:** funciones recursivas y condición de finalización.

```python
# Escriba su solución aquí
def cuenta_regresiva(numero):
    if numero < 1:
        print("El número debe ser mayor o igual a 1.")
        return
    if numero == 1:          # Caso base: detiene la recursividad
        print(1)
        print("¡Inicio!")
        return
    print(numero)
    cuenta_regresiva(numero - 1)   # Llamada recursiva con el número reducido

cuenta_regresiva(5)
print("---")
cuenta_regresiva(0)
```

### Ejercicio 10 — Sistema de ventas escolar (7 puntos)

Desarrolle un pequeño sistema utilizando **como mínimo cuatro funciones**:

1. `calcular_subtotal(precio, cantidad)`: retorna el subtotal.
2. `calcular_descuento(subtotal, porcentaje=0)`: retorna el monto descontado.
3. `calcular_total(*subtotales)`: retorna la suma de todos los subtotales recibidos.
4. `mostrar_factura(cliente, **detalles)`: muestra el cliente y los detalles finales de la compra.

Utilice estos productos:

| Producto | Precio (Gs.) | Cantidad |
|---|---:|---:|
| Cuaderno | 18.000 | 2 |
| Regla | 5.000 | 3 |
| Mochila | 120.000 | 1 |

Aplique un descuento del 10 % al total general. La factura debe mostrar cliente, subtotal general, descuento y total final. Formatee los montos con separador de miles.

**Contenido aplicado:** integración de funciones, `return`, parámetros predeterminados, `*args`, `**kwargs` y funciones integradas.

```python
# Datos disponibles
productos = [
    {"nombre": "Cuaderno", "precio": 18000, "cantidad": 2},
    {"nombre": "Regla", "precio": 5000, "cantidad": 3},
    {"nombre": "Mochila", "precio": 120000, "cantidad": 1}
]

# Escriba su solución aquí
def calcular_subtotal(precio, cantidad):
    return precio * cantidad
#calculamos el descuento
def calcular_descuento(subtotal, porcentaje=0):
    return subtotal * porcentaje / 100
#calculamos subtotales
def calcular_total(*subtotales):
    return sum(subtotales)
#funcion para mostrar facturas
def mostrar_factura(cliente, **detalles):
    print("========== FACTURA ==========")
    print(f"Cliente: {cliente}")
    for clave, valor in detalles.items():
        print(f"{clave}: {valor}")
    print("==============================")

def formatear_gs(monto):
    return f"Gs. {monto:,.0f}".replace(",", ".")

subtotales = [calcular_subtotal(p["precio"], p["cantidad"]) for p in productos]

subtotal_general = calcular_total(*subtotales)
descuento = calcular_descuento(subtotal_general, 10)
total_final = subtotal_general - descuento
#Mostramos la factura
mostrar_factura(
    "Fabian",
    Subtotal_general=formatear_gs(subtotal_general),
    Descuento_10_por_ciento=formatear_gs(descuento),
    Total_final=formatear_gs(total_final)
)
```
